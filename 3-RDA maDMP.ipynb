# 2. Pydantic model

```python
from __future__ import annotations

from typing import List, Optional
from pydantic import BaseModel, Field



# Common small reusable models

class IdentifierType(BaseModel):
    identifier: str = ""
    type: str = ""


class Affiliation(BaseModel):
    name: str = ""
    affiliation_id: Optional[IdentifierType] = None


class License(BaseModel):
    license_ref: str = ""
    start_date: str = ""


class RelatedIdentifier(BaseModel):
    identifier: str = ""
    type: str = ""
    relation_type: str = ""
    metadata_scheme: str = ""
    scheme_uri: str = ""
    scheme_type: str = ""
    resource_type: str = ""


class SecurityAndPrivacy(BaseModel):
    title: str = ""
    description: str = ""


class TechnicalResource(BaseModel):
    name: str = ""
    description: str = ""
    technical_resource_id: List[IdentifierType] = Field(default_factory=list)


class Metadata(BaseModel):
    description: str = ""
    language: str = ""
    metadata_standard_id: List[IdentifierType] = Field(default_factory=list)


class Host(BaseModel):
    title: str = ""
    url: str = ""
    description: str = ""
    availability: str = ""
    backup_frequency: str = ""
    backup_type: str = ""
    certified_with: str = ""
    geo_location: str = ""
    pid_system: List[str] = Field(default_factory=list)
    storage_type: str = ""
    support_versioning: str = ""
    host_id: List[IdentifierType] = Field(default_factory=list)


class Distribution(BaseModel):
    title: str = ""
    description: str = ""
    access_url: str = ""
    download_url: str = ""
    format: List[str] = Field(default_factory=list)
    byte_size: int = 0
    data_access: str = ""
    issued: str = ""
    available_until: str = ""
    license: List[License] = Field(default_factory=list)
    host: Optional[Host] = None


class Person(BaseModel):
    name: str = ""
    mbox: str = ""
    affiliation: List[Affiliation] = Field(default_factory=list)


class Contact(Person):
    contact_id: List[IdentifierType] = Field(default_factory=list)


class Contributor(Person):
    role: List[str] = Field(default_factory=list)
    contributor_id: List[IdentifierType] = Field(default_factory=list)


class Creator(Person):
    creator_id: List[IdentifierType] = Field(default_factory=list)


class Dataset(BaseModel):
    title: str = ""
    description: str = ""
    type: str = ""
    issued: str = ""
    language: str = ""
    keyword: List[str] = Field(default_factory=list)
    is_reused: bool = False
    personal_data: str = ""
    sensitive_data: str = ""
    preservation_statement: str = ""
    data_quality_assurance: List[str] = Field(default_factory=list)
    security_and_privacy: List[SecurityAndPrivacy] = Field(default_factory=list)
    rights: str = ""
    dataset_id: IdentifierType = Field(default_factory=IdentifierType)
    alternate_identifier: List[IdentifierType] = Field(default_factory=list)
    creator: List[Creator] = Field(default_factory=list)
    distribution: List[Distribution] = Field(default_factory=list)
    metadata: List[Metadata] = Field(default_factory=list)
    related_identifier: List[RelatedIdentifier] = Field(default_factory=list)
    technical_resource: List[TechnicalResource] = Field(default_factory=list)


class Cost(BaseModel):
    title: str = ""
    description: str = ""
    value: float = 0
    currency_code: str = ""


class Funding(BaseModel):
    funder_id: Optional[IdentifierType] = None
    grant_id: Optional[IdentifierType] = None
    funding_status: str = ""


class Project(BaseModel):
    title: str = ""
    description: str = ""
    start: str = ""
    end: str = ""
    project_id: List[IdentifierType] = Field(default_factory=list)
    funding: List[Funding] = Field(default_factory=list)


class DMP(BaseModel):
    dmp_id: IdentifierType = Field(default_factory=IdentifierType)
    title: str = ""
    description: str = ""
    created: str = ""
    modified: str = ""
    language: str = ""
    ethical_issues_exist: str = ""
    ethical_issues_description: str = ""
    ethical_issues_report: str = ""
    contact: Contact = Field(default_factory=Contact)
    contributor: List[Contributor] = Field(default_factory=list)
    cost: List[Cost] = Field(default_factory=list)
    project: List[Project] = Field(default_factory=list)
    dataset: List[Dataset] = Field(default_factory=list)
    alternate_identifier: List[IdentifierType] = Field(default_factory=list)
    related_identifier: List[RelatedIdentifier] = Field(default_factory=list)


class DMPDocument(BaseModel):
    dmp: DMP = Field(default_factory=DMP)
```

```python
import json

schema = DMPDocument.model_json_schema()

with open("rda_dmp_schema.json", "w", encoding="utf-8") as f:
    json.dump(schema, f, indent=2, ensure_ascii=False)

print("Schema saved to rda_dmp_schema.json")
```
