```python
import json
import pandas as pd
from copy import deepcopy


# -----------------------------
# 1. Empty maDMP skeleton
# -----------------------------
dmp_schema_skeleton = {
    "$schema": "./JSON-schema/1.2/maDMP-schema-1.2.json",
    "dmp": {
        "dmp_id": {
            "identifier": "",
            "type": ""
        },
        "title": "",
        "description": "",
        "created": "",
        "modified": "",
        "language": "",
        "ethical_issues_exist": "",
        "ethical_issues_description": "",
        "ethical_issues_report": "",
        "contact": {
            "name": "",
            "mbox": "",
            "contact_id": {
                "identifier": "",
                "type": ""
            },
            "affiliation": [
                {
                    "name": "",
                    "affiliation_id": {
                        "identifier": "",
                        "type": ""
                    }
                }
            ]
        },
        "contributor": [],
        "cost": [],
        "project": [],
        "dataset": [],
        "alternate_identifier": [],
        "related_identifier": []
    }
}


# -----------------------------
# 2. Helper templates
# -----------------------------
def new_affiliation():
    return {
        "name": "",
        "affiliation_id": {
            "identifier": "",
            "type": ""
        }
    }


def new_contributor():
    return {
        "name": "",
        "mbox": "",
        "role": [],
        "contributor_id": {
            "identifier": "",
            "type": ""
        },
        "affiliation": []
    }


def new_cost():
    return {
        "title": "",
        "description": "",
        "value": 0,
        "currency_code": ""
    }


def new_project():
    return {
        "title": "",
        "description": "",
        "start": "",
        "end": "",
        "project_id": [],
        "funding": []
    }


def new_project_id():
    return {
        "identifier": "",
        "type": ""
    }


def new_funding():
    return {
        "funder_name": "",
        "funder_id": {
            "identifier": "",
            "type": ""
        },
        "grant_id": {
            "identifier": "",
            "type": ""
        },
        "funding_status": ""
    }


def new_dataset():
    return {
        "dataset_id": {
            "identifier": "",
            "type": ""
        },
        "title": "",
        "description": "",
        "type": "",
        "issued": "",
        "language": "",
        "keyword": [],
        "is_reused": False,
        "personal_data": "",
        "sensitive_data": "",
        "preservation_statement": "",
        "data_quality_assurance": [],
        "security_and_privacy": [],
        "rights": "",
        "alternate_identifier": [],
        "creator": [],
        "distribution": [],
        "metadata": [],
        "related_identifier": [],
        "technical_resource": []
    }


def new_distribution():
    return {
        "title": "",
        "description": "",
        "access_url": "",
        "available_until": "",
        "byte_size": 0,
        "data_access": "",
        "download_url": "",
        "format": [],
        "issued": "",
        "license": [],
        "host": {
            "title": "",
            "url": "",
            "description": "",
            "availability": "",
            "backup_frequency": "",
            "backup_type": "",
            "certified_with": "",
            "geo_location": "",
            "host_id": {
                "identifier": "",
                "type": ""
            },
            "pid_system": [],
            "storage_type": "",
            "support_versioning": ""
        }
    }


def new_license():
    return {
        "license_ref": "",
        "start_date": ""
    }


def new_metadata():
    return {
        "description": "",
        "language": "",
        "metadata_standard_id": []
    }


def new_identifier():
    return {
        "identifier": "",
        "type": ""
    }


def new_related_identifier():
    return {
        "identifier": "",
        "type": "",
        "relation_type": "",
        "metadata_scheme": "",
        "scheme_uri": "",
        "scheme_type": "",
        "resource_type": ""
    }


def new_security_privacy():
    return {
        "title": "",
        "description": ""
    }


def new_creator():
    return {
        "name": "",
        "mbox": "",
        "creator_id": {
            "identifier": "",
            "type": ""
        },
        "affiliation": []
    }


def new_technical_resource():
    return {
        "name": "",
        "description": "",
        "technical_resource_id": []
    }


# -----------------------------
# 3. Value conversion helpers
# -----------------------------
def clean_value(value):
    if pd.isna(value):
        return ""
    return str(value).strip()


def parse_bool(value):
    val = clean_value(value).lower()
    if val in {"true", "yes", "1"}:
        return True
    if val in {"false", "no", "0"}:
        return False
    return False


def parse_number(value):
    val = clean_value(value)
    if val == "":
        return 0
    try:
        if "." in val:
            return float(val)
        return int(val)
    except ValueError:
        return val


def append_if_not_empty(lst, value):
    value = clean_value(value)
    if value:
        lst.append(value)


# -----------------------------
# 4. Main converter
# -----------------------------
def excel_to_madmp_json(excel_path, sheet_name=0):
    df = pd.read_excel(excel_path, sheet_name=sheet_name)

    required_cols = {"Section", "Name", "Value"}
    missing = required_cols - set(df.columns)
    if missing:
        raise ValueError(
            f"Missing required columns: {missing}. "
            f"Please add at least: Section, Name, Value"
        )

    # Group_ID is strongly recommended
    if "Group_ID" not in df.columns:
        df["Group_ID"] = ""

    result = deepcopy(dmp_schema_skeleton)

    # trackers for repeated blocks
    contributors = {}
    projects = {}
    fundings = {}
    datasets = {}
    distributions = {}
    metadata_blocks = {}
    licenses = {}
    creators = {}
    security_blocks = {}
    technical_resources = {}
    related_ids = {}
    alt_ids = {}

    for _, row in df.iterrows():
        section = clean_value(row["Section"]).lower()
        name = clean_value(row["Name"])
        value = row["Value"]
        group_id = clean_value(row["Group_ID"])

        if clean_value(value) == "":
            continue

        # -----------------
        # dmp level
        # -----------------
        if section == "dmp":
            if name == "title":
                result["dmp"]["title"] = clean_value(value)
            elif name == "description":
                result["dmp"]["description"] = clean_value(value)
            elif name == "created":
                result["dmp"]["created"] = clean_value(value)
            elif name == "modified":
                result["dmp"]["modified"] = clean_value(value)
            elif name == "language":
                result["dmp"]["language"] = clean_value(value)
            elif name == "ethical_issues_exist":
                result["dmp"]["ethical_issues_exist"] = clean_value(value)
            elif name == "ethical_issues_description":
                result["dmp"]["ethical_issues_description"] = clean_value(value)
            elif name == "ethical_issues_report":
                result["dmp"]["ethical_issues_report"] = clean_value(value)

        elif section == "dmp_id":
            if name == "identifier":
                result["dmp"]["dmp_id"]["identifier"] = clean_value(value)
            elif name == "type":
                result["dmp"]["dmp_id"]["type"] = clean_value(value)

        # -----------------
        # contact
        # -----------------
        elif section == "contact":
            if name == "name":
                result["dmp"]["contact"]["name"] = clean_value(value)
            elif name == "mbox":
                result["dmp"]["contact"]["mbox"] = clean_value(value)

        elif section == "contact_id":
            if name == "identifier":
                result["dmp"]["contact"]["contact_id"]["identifier"] = clean_value(value)
            elif name == "type":
                result["dmp"]["contact"]["contact_id"]["type"] = clean_value(value)

        elif section == "affiliation" and group_id == "CONTACT":
            if not result["dmp"]["contact"]["affiliation"]:
                result["dmp"]["contact"]["affiliation"].append(new_affiliation())
            aff = result["dmp"]["contact"]["affiliation"][0]
            if name == "name":
                aff["name"] = clean_value(value)

        elif section == "affiliation_id" and group_id == "CONTACT":
            if not result["dmp"]["contact"]["affiliation"]:
                result["dmp"]["contact"]["affiliation"].append(new_affiliation())
            aff = result["dmp"]["contact"]["affiliation"][0]
            if name == "identifier":
                aff["affiliation_id"]["identifier"] = clean_value(value)
            elif name == "type":
                aff["affiliation_id"]["type"] = clean_value(value)

        # -----------------
        # contributor
        # -----------------
        elif section == "contributor":
            if group_id not in contributors:
                contributors[group_id] = new_contributor()
                result["dmp"]["contributor"].append(contributors[group_id])

            obj = contributors[group_id]
            if name == "name":
                obj["name"] = clean_value(value)
            elif name == "mbox":
                obj["mbox"] = clean_value(value)
            elif name == "role":
                append_if_not_empty(obj["role"], value)

        elif section == "contributor_id":
            if group_id not in contributors:
                contributors[group_id] = new_contributor()
                result["dmp"]["contributor"].append(contributors[group_id])
            obj = contributors[group_id]
            if name == "identifier":
                obj["contributor_id"]["identifier"] = clean_value(value)
            elif name == "type":
                obj["contributor_id"]["type"] = clean_value(value)

        # -----------------
        # cost
        # -----------------
        elif section == "cost":
            if group_id not in result["dmp"]:
                pass
            if group_id not in alt_ids:
                pass
            if group_id not in fundings:
                pass

            cost_map = {c.get("_group_id", ""): c for c in result["dmp"]["cost"]}
            if group_id not in cost_map:
                obj = new_cost()
                obj["_group_id"] = group_id
                result["dmp"]["cost"].append(obj)
            else:
                obj = cost_map[group_id]

            if name == "title":
                obj["title"] = clean_value(value)
            elif name == "description":
                obj["description"] = clean_value(value)
            elif name == "value":
                obj["value"] = parse_number(value)
            elif name == "currency_code":
                obj["currency_code"] = clean_value(value)

        # -----------------
        # project
        # -----------------
        elif section == "project":
            if group_id not in projects:
                projects[group_id] = new_project()
                result["dmp"]["project"].append(projects[group_id])

            obj = projects[group_id]
            if name == "title":
                obj["title"] = clean_value(value)
            elif name == "description":
                obj["description"] = clean_value(value)
            elif name == "start":
                obj["start"] = clean_value(value)
            elif name == "end":
                obj["end"] = clean_value(value)

        elif section == "project_id":
            if group_id not in projects:
                projects[group_id] = new_project()
                result["dmp"]["project"].append(projects[group_id])

            obj = projects[group_id]
            if not obj["project_id"]:
                obj["project_id"].append(new_project_id())

            pid = obj["project_id"][0]
            if name == "identifier":
                pid["identifier"] = clean_value(value)
            elif name == "type":
                pid["type"] = clean_value(value)

        elif section == "funding":
            # funding row must use Group_ID like FUND1 and have Related_Project in Notes or use project group embedded
            # simplest: use Project_Group column if available; otherwise attach to first project
            project_group = clean_value(row["Project_Group"]) if "Project_Group" in df.columns else ""
            if project_group == "":
                if not projects:
                    projects["PROJ1"] = new_project()
                    result["dmp"]["project"].append(projects["PROJ1"])
                project_group = list(projects.keys())[0]

            if group_id not in fundings:
                fundings[group_id] = new_funding()
                projects[project_group]["funding"].append(fundings[group_id])

            obj = fundings[group_id]
            if name == "funder_name":
                obj["funder_name"] = clean_value(value)
            elif name == "funding_status":
                obj["funding_status"] = clean_value(value)

        elif section == "funder_id":
            project_group = clean_value(row["Project_Group"]) if "Project_Group" in df.columns else ""
            funding_group = group_id
            if project_group == "":
                if not projects:
                    projects["PROJ1"] = new_project()
                    result["dmp"]["project"].append(projects["PROJ1"])
                project_group = list(projects.keys())[0]
            if funding_group not in fundings:
                fundings[funding_group] = new_funding()
                projects[project_group]["funding"].append(fundings[funding_group])

            obj = fundings[funding_group]["funder_id"]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

        elif section == "grant_id":
            project_group = clean_value(row["Project_Group"]) if "Project_Group" in df.columns else ""
            funding_group = group_id
            if project_group == "":
                if not projects:
                    projects["PROJ1"] = new_project()
                    result["dmp"]["project"].append(projects["PROJ1"])
                project_group = list(projects.keys())[0]
            if funding_group not in fundings:
                fundings[funding_group] = new_funding()
                projects[project_group]["funding"].append(fundings[funding_group])

            obj = fundings[funding_group]["grant_id"]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

        # -----------------
        # dataset
        # -----------------
        elif section == "dataset":
            if group_id not in datasets:
                datasets[group_id] = new_dataset()
                result["dmp"]["dataset"].append(datasets[group_id])

            obj = datasets[group_id]
            if name == "title":
                obj["title"] = clean_value(value)
            elif name == "description":
                obj["description"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)
            elif name == "issued":
                obj["issued"] = clean_value(value)
            elif name == "language":
                obj["language"] = clean_value(value)
            elif name == "keyword":
                append_if_not_empty(obj["keyword"], value)
            elif name == "is_reused":
                obj["is_reused"] = parse_bool(value)
            elif name == "personal_data":
                obj["personal_data"] = clean_value(value)
            elif name == "sensitive_data":
                obj["sensitive_data"] = clean_value(value)
            elif name == "preservation_statement":
                obj["preservation_statement"] = clean_value(value)
            elif name == "data_quality_assurance":
                append_if_not_empty(obj["data_quality_assurance"], value)
            elif name == "rights":
                obj["rights"] = clean_value(value)

        elif section == "dataset_id":
            if group_id not in datasets:
                datasets[group_id] = new_dataset()
                result["dmp"]["dataset"].append(datasets[group_id])

            obj = datasets[group_id]["dataset_id"]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

        # -----------------
        # distribution
        # -----------------
        elif section == "distribution":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id

            if dataset_group not in datasets:
                datasets[dataset_group] = new_dataset()
                result["dmp"]["dataset"].append(datasets[dataset_group])

            if group_id not in distributions:
                distributions[group_id] = new_distribution()
                datasets[dataset_group]["distribution"].append(distributions[group_id])

            obj = distributions[group_id]
            if name == "title":
                obj["title"] = clean_value(value)
            elif name == "description":
                obj["description"] = clean_value(value)
            elif name == "access_url":
                obj["access_url"] = clean_value(value)
            elif name == "available_until":
                obj["available_until"] = clean_value(value)
            elif name == "byte_size":
                obj["byte_size"] = parse_number(value)
            elif name == "data_access":
                obj["data_access"] = clean_value(value)
            elif name == "download_url":
                obj["download_url"] = clean_value(value)
            elif name == "format":
                append_if_not_empty(obj["format"], value)
            elif name == "issued":
                obj["issued"] = clean_value(value)

        elif section == "host":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id
            if group_id not in distributions:
                distributions[group_id] = new_distribution()
                if dataset_group not in datasets:
                    datasets[dataset_group] = new_dataset()
                    result["dmp"]["dataset"].append(datasets[dataset_group])
                datasets[dataset_group]["distribution"].append(distributions[group_id])

            obj = distributions[group_id]["host"]
            if name == "title":
                obj["title"] = clean_value(value)
            elif name == "url":
                obj["url"] = clean_value(value)
            elif name == "description":
                obj["description"] = clean_value(value)
            elif name == "availability":
                obj["availability"] = clean_value(value)
            elif name == "backup_frequency":
                obj["backup_frequency"] = clean_value(value)
            elif name == "backup_type":
                obj["backup_type"] = clean_value(value)
            elif name == "certified_with":
                obj["certified_with"] = clean_value(value)
            elif name == "geo_location":
                obj["geo_location"] = clean_value(value)
            elif name == "pid_system":
                append_if_not_empty(obj["pid_system"], value)
            elif name == "storage_type":
                obj["storage_type"] = clean_value(value)
            elif name == "support_versioning":
                obj["support_versioning"] = clean_value(value)

        elif section == "host_id":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id
            if group_id not in distributions:
                distributions[group_id] = new_distribution()
                if dataset_group not in datasets:
                    datasets[dataset_group] = new_dataset()
                    result["dmp"]["dataset"].append(datasets[dataset_group])
                datasets[dataset_group]["distribution"].append(distributions[group_id])

            obj = distributions[group_id]["host"]["host_id"]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

        # -----------------
        # license
        # -----------------
        elif section == "license":
            distribution_group = clean_value(row["Distribution_Group"]) if "Distribution_Group" in df.columns else ""
            if distribution_group == "":
                distribution_group = group_id

            if distribution_group not in distributions:
                continue

            if group_id not in licenses:
                licenses[group_id] = new_license()
                distributions[distribution_group]["license"].append(licenses[group_id])

            obj = licenses[group_id]
            if name == "license_ref":
                obj["license_ref"] = clean_value(value)
            elif name == "start_date":
                obj["start_date"] = clean_value(value)

        # -----------------
        # metadata
        # -----------------
        elif section == "metadata":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id

            if dataset_group not in datasets:
                datasets[dataset_group] = new_dataset()
                result["dmp"]["dataset"].append(datasets[dataset_group])

            if group_id not in metadata_blocks:
                metadata_blocks[group_id] = new_metadata()
                datasets[dataset_group]["metadata"].append(metadata_blocks[group_id])

            obj = metadata_blocks[group_id]
            if name == "description":
                obj["description"] = clean_value(value)
            elif name == "language":
                obj["language"] = clean_value(value)

        elif section == "metadata_standard_id":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id

            metadata_group = group_id
            if metadata_group not in metadata_blocks:
                metadata_blocks[metadata_group] = new_metadata()
                datasets[dataset_group]["metadata"].append(metadata_blocks[metadata_group])

            if not metadata_blocks[metadata_group]["metadata_standard_id"]:
                metadata_blocks[metadata_group]["metadata_standard_id"].append(new_identifier())

            obj = metadata_blocks[metadata_group]["metadata_standard_id"][0]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

        # -----------------
        # security_and_privacy
        # -----------------
        elif section == "security_and_privacy":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id

            if dataset_group not in datasets:
                datasets[dataset_group] = new_dataset()
                result["dmp"]["dataset"].append(datasets[dataset_group])

            if group_id not in security_blocks:
                security_blocks[group_id] = new_security_privacy()
                datasets[dataset_group]["security_and_privacy"].append(security_blocks[group_id])

            obj = security_blocks[group_id]
            if name == "title":
                obj["title"] = clean_value(value)
            elif name == "description":
                obj["description"] = clean_value(value)

        # -----------------
        # creator
        # -----------------
        elif section == "creator":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id

            if dataset_group not in datasets:
                datasets[dataset_group] = new_dataset()
                result["dmp"]["dataset"].append(datasets[dataset_group])

            if group_id not in creators:
                creators[group_id] = new_creator()
                datasets[dataset_group]["creator"].append(creators[group_id])

            obj = creators[group_id]
            if name == "name":
                obj["name"] = clean_value(value)
            elif name == "mbox":
                obj["mbox"] = clean_value(value)

        elif section == "creator_id":
            creator_group = group_id
            if creator_group not in creators:
                continue
            obj = creators[creator_group]["creator_id"]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

        # -----------------
        # technical_resource
        # -----------------
        elif section == "technical_resource":
            dataset_group = clean_value(row["Dataset_Group"]) if "Dataset_Group" in df.columns else ""
            if dataset_group == "":
                dataset_group = group_id.split("_")[0] if "_" in group_id else group_id

            if dataset_group not in datasets:
                datasets[dataset_group] = new_dataset()
                result["dmp"]["dataset"].append(datasets[dataset_group])

            if group_id not in technical_resources:
                technical_resources[group_id] = new_technical_resource()
                datasets[dataset_group]["technical_resource"].append(technical_resources[group_id])

            obj = technical_resources[group_id]
            if name == "name":
                obj["name"] = clean_value(value)
            elif name == "description":
                obj["description"] = clean_value(value)

        elif section == "technical_resource_id":
            tr_group = group_id
            if tr_group not in technical_resources:
                continue
            if not technical_resources[tr_group]["technical_resource_id"]:
                technical_resources[tr_group]["technical_resource_id"].append(new_identifier())
            obj = technical_resources[tr_group]["technical_resource_id"][0]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

        # -----------------
        # related_identifier
        # -----------------
        elif section == "related_identifier":
            # root-level related ids
            if group_id not in related_ids:
                related_ids[group_id] = new_related_identifier()
                result["dmp"]["related_identifier"].append(related_ids[group_id])

            obj = related_ids[group_id]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)
            elif name == "relation_type":
                obj["relation_type"] = clean_value(value)
            elif name == "metadata_scheme":
                obj["metadata_scheme"] = clean_value(value)
            elif name == "scheme_uri":
                obj["scheme_uri"] = clean_value(value)
            elif name == "scheme_type":
                obj["scheme_type"] = clean_value(value)
            elif name == "resource_type":
                obj["resource_type"] = clean_value(value)

        # -----------------
        # alternate_identifier
        # -----------------
        elif section == "alternate_identifier":
            if group_id not in alt_ids:
                alt_ids[group_id] = new_identifier()
                result["dmp"]["alternate_identifier"].append(alt_ids[group_id])

            obj = alt_ids[group_id]
            if name == "identifier":
                obj["identifier"] = clean_value(value)
            elif name == "type":
                obj["type"] = clean_value(value)

    # cleanup helper fields
    for cost in result["dmp"]["cost"]:
        cost.pop("_group_id", None)

    return result


# -----------------------------
# 5. Save JSON
# -----------------------------
def save_json(data, output_path):
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2, ensure_ascii=False)


# -----------------------------
# 6. Example usage
# -----------------------------
if __name__ == "__main__":
    excel_file = "filled_madmp_extraction.xlsx"   # your edited Excel file
    output_json = "filled_madmp.json"

    madmp_json = excel_to_madmp_json(excel_file)
    save_json(madmp_json, output_json)

    print(f"Saved JSON to: {output_json}")
```

# 2. Pydantic model

```python
from __future__ import annotations

from typing import List, Optional
from pydantic import BaseModel, Field



# Common small reusable models

class IdentifierType(BaseModel):
    identifier: str = ""
    type: str = ""


class Affiliation(BaseModel):
    name: str = ""
    affiliation_id: Optional[IdentifierType] = None


class License(BaseModel):
    license_ref: str = ""
    start_date: str = ""


class RelatedIdentifier(BaseModel):
    identifier: str = ""
    type: str = ""
    relation_type: str = ""
    metadata_scheme: str = ""
    scheme_uri: str = ""
    scheme_type: str = ""
    resource_type: str = ""


class SecurityAndPrivacy(BaseModel):
    title: str = ""
    description: str = ""


class TechnicalResource(BaseModel):
    name: str = ""
    description: str = ""
    technical_resource_id: List[IdentifierType] = Field(default_factory=list)


class Metadata(BaseModel):
    description: str = ""
    language: str = ""
    metadata_standard_id: List[IdentifierType] = Field(default_factory=list)


class Host(BaseModel):
    title: str = ""
    url: str = ""
    description: str = ""
    availability: str = ""
    backup_frequency: str = ""
    backup_type: str = ""
    certified_with: str = ""
    geo_location: str = ""
    pid_system: List[str] = Field(default_factory=list)
    storage_type: str = ""
    support_versioning: str = ""
    host_id: List[IdentifierType] = Field(default_factory=list)


class Distribution(BaseModel):
    title: str = ""
    description: str = ""
    access_url: str = ""
    download_url: str = ""
    format: List[str] = Field(default_factory=list)
    byte_size: int = 0
    data_access: str = ""
    issued: str = ""
    available_until: str = ""
    license: List[License] = Field(default_factory=list)
    host: Optional[Host] = None


class Person(BaseModel):
    name: str = ""
    mbox: str = ""
    affiliation: List[Affiliation] = Field(default_factory=list)


class Contact(Person):
    contact_id: List[IdentifierType] = Field(default_factory=list)


class Contributor(Person):
    role: List[str] = Field(default_factory=list)
    contributor_id: List[IdentifierType] = Field(default_factory=list)


class Creator(Person):
    creator_id: List[IdentifierType] = Field(default_factory=list)


class Dataset(BaseModel):
    title: str = ""
    description: str = ""
    type: str = ""
    issued: str = ""
    language: str = ""
    keyword: List[str] = Field(default_factory=list)
    is_reused: bool = False
    personal_data: str = ""
    sensitive_data: str = ""
    preservation_statement: str = ""
    data_quality_assurance: List[str] = Field(default_factory=list)
    security_and_privacy: List[SecurityAndPrivacy] = Field(default_factory=list)
    rights: str = ""
    dataset_id: IdentifierType = Field(default_factory=IdentifierType)
    alternate_identifier: List[IdentifierType] = Field(default_factory=list)
    creator: List[Creator] = Field(default_factory=list)
    distribution: List[Distribution] = Field(default_factory=list)
    metadata: List[Metadata] = Field(default_factory=list)
    related_identifier: List[RelatedIdentifier] = Field(default_factory=list)
    technical_resource: List[TechnicalResource] = Field(default_factory=list)


class Cost(BaseModel):
    title: str = ""
    description: str = ""
    value: float = 0
    currency_code: str = ""


class Funding(BaseModel):
    funder_id: Optional[IdentifierType] = None
    grant_id: Optional[IdentifierType] = None
    funding_status: str = ""


class Project(BaseModel):
    title: str = ""
    description: str = ""
    start: str = ""
    end: str = ""
    project_id: List[IdentifierType] = Field(default_factory=list)
    funding: List[Funding] = Field(default_factory=list)


class DMP(BaseModel):
    dmp_id: IdentifierType = Field(default_factory=IdentifierType)
    title: str = ""
    description: str = ""
    created: str = ""
    modified: str = ""
    language: str = ""
    ethical_issues_exist: str = ""
    ethical_issues_description: str = ""
    ethical_issues_report: str = ""
    contact: Contact = Field(default_factory=Contact)
    contributor: List[Contributor] = Field(default_factory=list)
    cost: List[Cost] = Field(default_factory=list)
    project: List[Project] = Field(default_factory=list)
    dataset: List[Dataset] = Field(default_factory=list)
    alternate_identifier: List[IdentifierType] = Field(default_factory=list)
    related_identifier: List[RelatedIdentifier] = Field(default_factory=list)


class DMPDocument(BaseModel):
    dmp: DMP = Field(default_factory=DMP)
```
